2. calcule `dx` e `dy` como as diferenças **em valor absoluto** entre as coordenadas;
3. calcule a distância entre os pontos com a fórmula abaixo:

$$distancia = \sqrt{dx^2 + dy^2}$$

4. calcule a área do círculo que tem essa distância como raio;
5. exiba `dx`, `dy`, a distância e a área, todos arredondados para 2 casas decimais com `round`.


````python
# Questão 3 - Distância entre dois pontos e área do círculo
## importando a biblioteca math para cálculos matemáticos
import math

## Entrada de dados
x1 = float(input("Informe x1: "))
y1 = float(input("Informe y1: "))
x2 = float(input("Informe x2: "))
y2 = float(input("Informe y2: "))


## Calculos e resultados (todos arredondados para 2 casas decimais)
distancia = round(math.sqrt((x2 - x1) ** 2 + (y2 - y1) ** 2), 2)

print("Dx = ", round(x2 - x1, 2))
print("Dy = ", round(y2 - y1, 2))
print("A distância entre os pontos é:", distancia)
print("A área do círculo de raio ", distancia, " é: ", round(math.pi * distancia ** 2, 2))

````

- imprimir o `texto` entre dois `caractere` (um no início e outro no fim da linha);
- imprimir novamente a linha de fechamento.

A função deve ter uma **docstring**. Note que ela **não retorna valor**, apenas exibe (retorna `None`).

Exemplo:

```

```

Saída:

```

********************
```

**(b)** Chame `moldura` **três vezes**, com textos, caracteres e larguras diferentes, mostrando que a função foi
**generalizada** pelos parâmetros.

````python
def moldura_texto(texto, caracatere, largura):

    # Calcula o comprimento do texto
    comprimento_texto = len(texto)

    # Cria a linha superior e inferior da moldura
    linha_superior_inferior = caracatere * largura

    # Calcula o espaço restante para centralizar o texto
    espaco_restante = largura - comprimento_texto - 2  # 2 para os espaços ao redor do texto
    espaco_esquerda = espaco_restante // 2
    espaco_direita = espaco_restante - espaco_esquerda

    # Cria a linha do meio com o texto centralizado
    linha_meio = f"{caracatere}{' ' * espaco_esquerda}{texto}{' ' * espaco_direita}{caracatere}"

    # Combina todas as linhas para formar a moldura completa
    moldura = f"{linha_superior_inferior}\n{linha_meio}\n{linha_superior_inferior}"

    return moldura

for i in range(3):
    print(moldura_texto("Trabalho", "*", 40))
    print(moldura_texto("Python", "#", 50))
    print(moldura_texto("Moldura", "=", 60))


    """
    Função que cria uma moldura ao redor de um texto usando um caractere específico.

    Parâmetros:
    texto (str): O texto a ser moldurado.
    caracatere (str): O caractere a ser usado para criar a moldura.
    largura (int): A largura total da moldura.

    Retorna:
    str: O texto moldurado com a moldura.
    """
````

```
eh_triangulo(-1, 4, 5) -> False
```

````python
def eh_bissexto(ano):
    if (ano % 4 == 0 and ano % 100 != 0) or (ano % 400 == 0):
        return True
    else:
        return False


def eh_triangulo(a, b, c):
    if a + b > c and a + c > b and b + c > a:
        return True
    else:
        return False


for ano in range(1996, 2006):
    if eh_bissexto(ano):
        print(f"{ano} é um ano bissexto.")
    else:
        print(f"{ano} não é um ano bissexto.")

print("Triangulos:")
print(eh_triangulo(3, 4, 5))
print(eh_triangulo(1, 2, 3))
print(eh_triangulo(5, 5, 5))
````

| IMC | Classificação |
|---|---|

| de 18,5 a 24,9 | `"Peso normal"` |
| de 25,0 a 29,9 | `"Sobrepeso"` |
| 30,0 ou mais | `"Obesidade"` |

````python

"""
essa função media_notas recebe as notas como parametro e retorna com o calculo da media
"""

def media_notas(nota1, nota2, nota3):
    return (nota1 + nota2 + nota3) / 3


""""
a def situação recebe como parametro a media e retorna se o aluno está aprovado, recuperação ou reprovado com base na media
"""

def situacao(media):
    if media >= 7:
        return "Aprovado"
    elif media >= 4:
        return "Recuperação"
    else:
        return "Reprovado"


qtd_alunos = int(input("Informe a quantidade de alunos da turma: "))
boletim_linhas = []
notas_turma = []
maior_media = -1
nome_melhor = ""

for i in range(qtd_alunos):
    print("\nAluno:")
    nome = input("nome: ")
    nota1 = float(input("Nota 1: "))
    nota2 = float(input("Nota 2: "))
    nota3 = float(input("Nota 3: "))

    m = media_notas(nota1, nota2, nota3)
    s = situacao(m)

    boletim_linhas.append(f"{nome}: {m:.1f} - {s}")
    notas_turma.append(m)

    # Compara a média atual com a maior registrada até agora
    if m > maior_media:
        maior_media = m
        nome_melhor = nome


print("\n===== BOLETIM =====")
for linha in boletim_linhas:
    print(linha)
print("=====================")

# ----------------------------

print("Media geral da turma: ", sum(notas_turma) / len(notas_turma))
print("Aprovados: ", len(notas_turma) - len([n for n in notas_turma if n < 7]), " de ", len(notas_turma), "(", round((len(notas_turma) - len([n for n in notas_turma if n < 7])) / len(notas_turma) * 100, 2), "%)")
print("Melhor média: ", nome_melhor, "(",max(notas_turma), ")")

````
